```python
import os
import pandas as pd
import numpy as np
from fingerprints import smiles_to_fingerprint
from training import run_pca_scatter, plot_pca_variance

# -----------------------------
# Wczytanie danych
# -----------------------------
df_balanced = pd.read_csv("df_to_RF.csv")
X_train_smiles = df_balanced["Canonicalized SMILES"].tolist()
y_train = df_balanced["Canonicalized Taste"].values

# -----------------------------
# SMILES -> fingerprinty
# -----------------------------
X_train = np.vstack([smiles_to_fingerprint(s) for s in X_train_smiles])

# -----------------------------
# Folder na wykresy
# -----------------------------
fig_folder = "figures"
os.makedirs(fig_folder, exist_ok=True)

# -----------------------------
# Wywołanie funkcji PCA
# -----------------------------
run_pca_scatter(X_train, y_train, save_path=f"{fig_folder}/pca_2d_scatter.png")
plot_pca_variance(X_train, n_components=10, save_path=f"{fig_folder}/pca_variance.png")
```
